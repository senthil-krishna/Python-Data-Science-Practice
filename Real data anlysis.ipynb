```python
import pandas as pd

data = {
    "Student": ["Arun", "Bala", "Priya", "Kumar", "Ravi"],
    "Hours": [2, 3, 5, 4, 6],
    "Marks": [45, 55, 85, 70, 95]
}

df = pd.DataFrame(data)

df.to_csv("students.csv", index=False)

print("CSV file created successfully!")

df = pd.read_csv("students.csv")
df

print("Average:", df["Marks"].mean())
print("Highest:", df["Marks"].max())
print("Lowest:", df["Marks"].min())
print("Average Hours:", df["Hours"].mean())
print(df["Student"])
high_score=df[df["Marks"] > 60]
print(high_score)
```

**Data cleaning**

```python
data = {
    "Student": ["Arun", "Bala", "Priya", "Kumar"],
    "Marks": [75, None, 90, None]
}

df = pd.DataFrame(data)

print(df.isnull().sum())
df["Marks"] = df["Marks"].fillna(0)

print(df)
```

**Missing value using fill method**

```python
data = {
    "Student": ["Arun", "Bala", "Priya", "Kumar"],
    "Marks": [75, None, 90, None]
}

df = pd.DataFrame(data)
average=df["Marks"].mean()
df["Marks"]=df["Marks"].fillna(average)
print(df)
print(average)
```

**Combine numpy,pandas and matplotlib**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "Student": ["Arun", "Bala", "Priya", "Kumar"],
    "Marks": [75, 82, 90, 65]
})

print("Average:", np.mean(df["Marks"]))

plt.bar(df["Student"], df["Marks"])
plt.title("Student Marks")
plt.show()
```
