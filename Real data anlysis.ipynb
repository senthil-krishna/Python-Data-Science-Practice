**Data cleaning**

```python
data = {
    "Student": ["Arun", "Bala", "Priya", "Kumar"],
    "Marks": [75, None, 90, None]
}

df = pd.DataFrame(data)

print(df.isnull().sum())
df["Marks"] = df["Marks"].fillna(0)

print(df)
```

**Missing value using fill method**

```python
data = {
    "Student": ["Arun", "Bala", "Priya", "Kumar"],
    "Marks": [75, None, 90, None]
}

df = pd.DataFrame(data)
average=df["Marks"].mean()
df["Marks"]=df["Marks"].fillna(average)
print(df)
print(average)
```

**Combine numpy,pandas and matplotlib**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "Student": ["Arun", "Bala", "Priya", "Kumar"],
    "Marks": [75, 82, 90, 65]
})

print("Average:", np.mean(df["Marks"]))

plt.bar(df["Student"], df["Marks"])
plt.title("Student Marks")
plt.show()
```

**Student performence analysis**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
df = pd.DataFrame({
    "Student": ["Arun", "Bala", "Priya", "Kumar"],
    "Marks": [75, 32, 90, 65]
})
df["Result"] = np.where(df["Marks"] >= 40, "Pass", "Fail")
print(df)
print("Pass:", (df["Marks"] >= 40).sum())
print("Fail:", (df["Marks"] < 40).sum())
df["Marks"].plot(kind="bar")
plt.title("Student Performance")
plt.ylabel("Marks")
plt.show()
```

```python
pass_count = (df["Marks"] >= 40).sum()
total_students = len(df)

pass_percentage = (pass_count / total_students) * 100

print("Pass Percentage:", pass_percentage, "%")
```

```python
print("Total Students:", len(df))
print("Average Marks:", df["Marks"].mean())
print("Highest Marks:", df["Marks"].max())
print("Lowest Marks:", df["Marks"].min())

top_student = df.loc[df["Marks"].idxmax(), "Student"]
print("Top Student:", top_student)
```
