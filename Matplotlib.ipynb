```python
import matplotlib.pyplot as plt
days = [1, 2, 3, 4, 5]
marks = [40, 50, 45, 70, 85]
plt.plot(days, marks)
plt.xlabel("Days")
plt.ylabel("Marks")
plt.title("Student Marks Progress")
plt.show()
```

**Bar chart**

```python
students = ["Arun", "Bala", "Priya", "Kumar"]
marks = [70, 85, 90, 60]
plt.bar(students,marks)
plt.xlabel("Students")
plt.ylabel("Marks")
plt.title("Student marks comparision")
plt.show()
```

```python
students = ["A", "B", "C"]
marks = [40, 75, 95]
plt.bar(students, marks)
plt.show()
```

**Pie chart**

```python
students = ["A", "B", "C"]
marks = [40, 75, 95]
plt.pie(marks,labels=students,autopct="%1.1f%%")
plt.title("Student marks share")
plt.show()
```

**Scatter plot**

```python
hours = [1, 2, 3, 4, 5]
marks = [35, 45, 60, 75, 90]
plt.scatter(hours, marks)
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks")
plt.show()
```
