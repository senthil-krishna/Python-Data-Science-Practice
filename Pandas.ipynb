**Pandas**

```python
import pandas as pd
data={
    "Name":["kumar","david","siva"],
    "Marks":[94,97,79]
}
df=pd.DataFrame(data)
print(df)
```

```python
data = {
    "Name": ["Ravi", "Priya", "Mani"],
    "Age": [20, 21, 19]
}
df = pd.DataFrame(data)
print(df["Name"])
print(df["Age"].mean())
```

**Find the missing value using isnull()**

```python
data = {
    "Name": ["Ravi", "Priya", "Mani"],
    "Marks": [75, None, 85]
}
df = pd.DataFrame(data)
print(df.isnull().sum())
```

```python
import pandas as pd
data = {
    "Name": ["Arun", "Bala", "Kumar"],
    "Marks": [60, None, 90]
}
df = pd.DataFrame(data)
df["Marks"] = df["Marks"].fillna(75)#fillna means fill the missing values
print(df)
```
