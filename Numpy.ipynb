```python
import numpy as np
marks=np.array([10,20,30,40,50])
print(marks)
```

```python
marks=np.array([20,35,78,80,60])
print(marks+5)
print(marks-4)
print(marks*2)
print(marks/3)
```

**NumPy Important Functions**

```python
marks=np.array([70,80,90,60,85])
print(np.sum(marks))
print(np.mean(marks))
print(np.max(marks))
print(np.min(marks))
print(np.sort(marks))
```

**Array Indexing**

```python
marks=np.array([70,80,90,60,85])
print(marks[0])
print(marks[1])
print(marks[-1])
```

```python
import numpy as np
numbers=np.array([10,20,30,40,50])
print(numbers)
print(np.sum(numbers))
print(np.mean(numbers))
print(np.max(numbers))
print(numbers+10)
```

**NumPy Filtering**

```python
numbers = np.array([10, 25, 40, 55, 70])
print(numbers[numbers > 30])
print(numbers[numbers < 50])
```

**shape and reshape**

```python
data=np.array([10,20,30,40,50,60])
print(data.shape)
result=data.reshape(3,2)
print(result)
print(result.shape)
```

**Axis concept**

```python
data=np.array([[10,20],
              [30,40],
              [50,60]])
print(np.sum(data,axis=0))#column wise calculation
print(np.sum(data,axis=1))#Row wise calculation
```
