**Using numpy,pandas and matplotlib to generate student mark analysis**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
data = {
    "Student": ["Arun", "Bala", "Priya", "Kumar", "Ravi","Divya"],
    "Hours": [1, 2, 3, 4, 5,6],
    "Marks": [35, 45, 60, 75, 90,95]
}
df=pd.DataFrame(data)
print(df)
print("Higest Marks:",np.max(df["Marks"]))
print("Average marks:",np.mean(df["Marks"]))
plt.plot(df["Student"],df["Marks"])
plt.xlabel("Students")
plt.ylabel("Marks")
plt.title("Student mark analysis")
plt.show()
```
